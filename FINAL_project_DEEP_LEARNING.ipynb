```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from collections import Counter

train = pd.read_csv('data/train.csv')
test = pd.read_csv('data/test.csv')

print("=== AdoptionSpeed distribution ===")
print(train['AdoptionSpeed'].value_counts().sort_index())
print()

print("=== Description length ===")
train['desc_len'] = train['Description'].fillna('').str.len()
print(train['desc_len'].describe())
print()

# Скільки фото на одну тварину
import os
from collections import defaultdict

train_img_dir = 'data/images/images/train'
img_counts = defaultdict(int)

for fname in os.listdir(train_img_dir):
    if fname.endswith('.jpg'):
        pet_id = fname.split('-')[0]
        img_counts[pet_id] += 1

train['num_images'] = train['PetID'].map(img_counts).fillna(0).astype(int)

print("=== Number of images per pet ===")
print(train['num_images'].describe())
print()
print(train['num_images'].value_counts().sort_index().head(15))
```

```python
# TEST MODEL (TF-IDF + LightGBM)
!pip install -q lightgbm scikit-learn

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import cohen_kappa_score
import lightgbm as lgb
import numpy as np
import pandas as pd

# Підготовка тексту
train['Description'] = train['Description'].fillna('')
test['Description'] = test['Description'].fillna('')

# TF-IDF
tfidf = TfidfVectorizer(
    max_features=15000,
    ngram_range=(1, 2),
    stop_words='english',
    min_df=3
)

X_text = tfidf.fit_transform(train['Description'])
X_test_text = tfidf.transform(test['Description'])

y = train['AdoptionSpeed'].values

print("TF-IDF shape:", X_text.shape)
```

```python
def quadratic_weighted_kappa(y_true, y_pred):
    return cohen_kappa_score(y_true, y_pred, weights='quadratic')

# Крос-валідація
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

oof_preds = np.zeros(len(train))
n_classes = len(np.unique(y))
test_preds = np.zeros((len(test), n_classes))

print("Classes in data:", np.unique(y))
print("Number of classes:", n_classes)

for fold, (train_idx, val_idx) in enumerate(skf.split(X_text, y)):
    print(f"\nFold {fold+1}")

    X_tr, X_val = X_text[train_idx], X_text[val_idx]
    y_tr, y_val = y[train_idx], y[val_idx]

    model = lgb.LGBMClassifier(
        n_estimators=1000,
        learning_rate=0.05,
        num_leaves=64,
        max_depth=-1,
        subsample=0.8,
        colsample_bytree=0.8,
        random_state=42,
        n_jobs=-1,
        verbosity=-1
    )

    model.fit(
        X_tr, y_tr,
        eval_set=[(X_val, y_val)],
        eval_metric='multi_logloss',
        callbacks=[lgb.early_stopping(50, verbose=False)]
    )

    oof_preds[val_idx] = model.predict(X_val)
    test_preds += model.predict_proba(X_test_text) / 5

# Оцінка
qwk = quadratic_weighted_kappa(y, oof_preds)
print(f"\nOOF Quadratic Weighted Kappa: {qwk:.5f}")
```

```python
# ADD the quantity photos + optimize THRESHOLDS
from scipy.optimize import minimize

# Додаємо кількість зображень як фічу
from scipy.sparse import hstack, csr_matrix

num_images_train = train['num_images'].values.reshape(-1, 1)
num_images_test = test['PetID'].map(
    lambda x: len([f for f in os.listdir('data/images/images/test') if f.startswith(x)])
).fillna(0).values.reshape(-1, 1)

X_text_img = hstack([X_text, csr_matrix(num_images_train)])
X_test_text_img = hstack([X_test_text, csr_matrix(num_images_test)])

print("New shape:", X_text_img.shape)

# Перенавчаємо з новою фічею (швидко)
oof_preds = np.zeros(len(train))
test_preds = np.zeros((len(test), n_classes))

for fold, (train_idx, val_idx) in enumerate(skf.split(X_text_img, y)):
    print(f"Fold {fold+1}", end=' ')

    X_tr, X_val = X_text_img[train_idx], X_text_img[val_idx]
    y_tr, y_val = y[train_idx], y[val_idx]

    model = lgb.LGBMClassifier(
        n_estimators=800,
        learning_rate=0.05,
        num_leaves=64,
        subsample=0.8,
        colsample_bytree=0.8,
        random_state=42,
        n_jobs=-1,
        verbosity=-1
    )

    model.fit(X_tr, y_tr,
              eval_set=[(X_val, y_val)],
              callbacks=[lgb.early_stopping(40, verbose=False)])

    oof_preds[val_idx] = model.predict(X_val)
    test_preds += model.predict_proba(X_test_text_img) / 5

print(f"\nOOF QWK with num_images: {quadratic_weighted_kappa(y, oof_preds):.5f}")
```

```python
# OPTIMIZED Rounder
from sklearn.metrics import cohen_kappa_score
from scipy.optimize import minimize

class OptimizedRounder:
    def __init__(self):
        self.coef_ = None

    def _kappa_loss(self, coef, X, y):
        X_p = np.copy(X)
        for i, pred in enumerate(X_p):
            if pred < coef[0]:
                X_p[i] = 1
            elif pred < coef[1]:
                X_p[i] = 2
            elif pred < coef[2]:
                X_p[i] = 3
            else:
                X_p[i] = 4
        return -cohen_kappa_score(y, X_p, weights='quadratic')

    def fit(self, X, y):
        loss_partial = lambda coef: self._kappa_loss(coef, X, y)
        initial_coef = [1.5, 2.5, 3.5]
        self.coef_ = minimize(loss_partial, initial_coef, method='nelder-mead')['x']

    def predict(self, X, coef):
        X_p = np.copy(X)
        for i, pred in enumerate(X_p):
            if pred < coef[0]:
                X_p[i] = 1
            elif pred < coef[1]:
                X_p[i] = 2
            elif pred < coef[2]:
                X_p[i] = 3
            else:
                X_p[i] = 4
        return X_p.astype(int)

# Отримуємо "сирі" предикти (середнє по класах)
oof_raw = np.zeros(len(train))
test_raw = np.zeros(len(test))

# Перераховуємо з predict_proba (середнє)
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
oof_raw = np.zeros(len(train))
test_raw = np.zeros(len(test))

for fold, (train_idx, val_idx) in enumerate(skf.split(X_text_img, y)):
    X_tr, X_val = X_text_img[train_idx], X_text_img[val_idx]
    y_tr = y[train_idx]

    model = lgb.LGBMClassifier(
        n_estimators=800, learning_rate=0.05, num_leaves=64,
        subsample=0.8, colsample_bytree=0.8, random_state=42, n_jobs=-1, verbosity=-1
    )
    model.fit(X_tr, y_tr)

    # Використовуємо expected value
    proba_val = model.predict_proba(X_val)
    proba_test = model.predict_proba(X_test_text_img)

    classes = model.classes_
    oof_raw[val_idx] = proba_val @ classes
    test_raw += (proba_test @ classes) / 5

# Оптимізуємо пороги
optR = OptimizedRounder()
optR.fit(oof_raw, y)
print("Optimal coefficients:", optR.coef_)

oof_final = optR.predict(oof_raw, optR.coef_)
print(f"OOF QWK after threshold optimization: {quadratic_weighted_kappa(y, oof_final):.5f}")
```

```python
import torch
import timm
from torchvision import transforms
from PIL import Image
from tqdm.notebook import tqdm
import os
import numpy as np
from collections import defaultdict

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print("Device:", device)

# Модель
model = timm.create_model('efficientnet_b0', pretrained=True, num_classes=0)
model = model.to(device)
model.eval()

# Трансформації
transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
])

def get_pet_embeddings(image_dir, pet_ids):
    embeddings = {}

    for pet_id in tqdm(pet_ids, desc="Processing pets"):
        # Знаходимо всі фото цієї тварини
        files = [f for f in os.listdir(image_dir) if f.startswith(pet_id) and f.endswith('.jpg')]

        if len(files) == 0:
            embeddings[pet_id] = np.zeros(1280)  # розмір ембедінгу EfficientNet-B0
            continue

        pet_embs = []
        for f in files[:6]:  # максимум 6 фото на тварину (щоб швидше)
            try:
                img = Image.open(os.path.join(image_dir, f)).convert('RGB')
                img_tensor = transform(img).unsqueeze(0).to(device)

                with torch.no_grad():
                    emb = model(img_tensor).cpu().numpy().flatten()
                pet_embs.append(emb)
            except:
                continue

        if pet_embs:
            embeddings[pet_id] = np.mean(pet_embs, axis=0)
        else:
            embeddings[pet_id] = np.zeros(1280)

    return embeddings

# Витягуємо ембедінги
print("Extracting train embeddings...")
train_emb_dict = get_pet_embeddings('data/images/images/train', train['PetID'].values)

print("Extracting test embeddings...")
test_emb_dict = get_pet_embeddings('data/images/images/test', test['PetID'].values)

# Перетворюємо в матриці
train_img_emb = np.stack([train_emb_dict[pid] for pid in train['PetID']])
test_img_emb = np.stack([test_emb_dict[pid] for pid in test['PetID']])

print("Train embeddings shape:", train_img_emb.shape)
print("Test embeddings shape:", test_img_emb.shape)

# Зберігаємо, щоб не перераховувати
np.save('train_img_emb.npy', train_img_emb)
np.save('test_img_emb.npy', test_img_emb)
print("Embeddings saved!")
```

```python
# TRAIN The final model
from scipy.sparse import hstack, csr_matrix
from sklearn.decomposition import PCA

# Зменшуємо розмірність ембедінгів (щоб швидше і менше переобучення)
pca = PCA(n_components=64, random_state=42)
train_img_pca = pca.fit_transform(train_img_emb)
test_img_pca = pca.transform(test_img_emb)

print("PCA shape:", train_img_pca.shape)

# Збираємо всі фічі разом
X_final = hstack([
    X_text,
    csr_matrix(num_images_train),
    csr_matrix(train_img_pca)
]).tocsr()

X_test_final = hstack([
    X_test_text,
    csr_matrix(num_images_test),
    csr_matrix(test_img_pca)
]).tocsr()

print("Final feature shape:", X_final.shape)

# Тренуємо фінальну модель
oof_raw = np.zeros(len(train))
test_raw = np.zeros(len(test))

skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

for fold, (train_idx, val_idx) in enumerate(skf.split(X_final, y)):
    print(f"Fold {fold+1}")

    X_tr, X_val = X_final[train_idx], X_final[val_idx]
    y_tr, y_val = y[train_idx], y[val_idx]

    model = lgb.LGBMClassifier(
        n_estimators=1200,
        learning_rate=0.03,
        num_leaves=96,
        subsample=0.8,
        colsample_bytree=0.7,
        reg_alpha=0.1,
        reg_lambda=0.1,
        random_state=42,
        n_jobs=-1,
        verbosity=-1
    )

    model.fit(
        X_tr, y_tr,
        eval_set=[(X_val, y_val)],
        callbacks=[lgb.early_stopping(60, verbose=False)]
    )

    proba_val = model.predict_proba(X_val)
    proba_test = model.predict_proba(X_test_final)

    classes = model.classes_
    oof_raw[val_idx] = proba_val @ classes
    test_raw += (proba_test @ classes) / 5

# Оптимізуємо пороги
optR = OptimizedRounder()
optR.fit(oof_raw, y)
print("Optimal coefficients:", optR.coef_)

oof_final = optR.predict(oof_raw, optR.coef_)
final_qwk = quadratic_weighted_kappa(y, oof_final)
print(f"\nFinal OOF Quadratic Weighted Kappa: {final_qwk:.5f}")
```

```python
# SUBMISSION
# Фінальні предикти на тест
test_final = optR.predict(test_raw, optR.coef_)

submission = pd.DataFrame({
    'PetID': test['PetID'],
    'AdoptionSpeed': test_final.astype(int)
})

submission.to_csv('submission.csv', index=False)
print(submission.head(10))
print("\nSubmission shape:", submission.shape)
print("Distribution of predictions:")
print(submission['AdoptionSpeed'].value_counts().sort_index())
```
